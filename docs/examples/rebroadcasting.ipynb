## Key Concepts
- **DataSource**: An abstract base class for event sources. Subclass it and implement run().
- **RebroadcastClient**: Consumes events from a DataSource and sends them to ATProto with automatic retries and exponential backoff.
- **Thread safety**: Event producers and consumers communicate through a thread-safe queue.
